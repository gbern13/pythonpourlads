```python
import s3fs
import pandas as pd
import tempfile
fs = s3fs.S3FileSystem(client_kwargs={"endpoint_url": "https://minio.lab.sspcloud.fr"})

MY_BUCKET = "gbern13"

FILE_PATH_S3 = f"{MY_BUCKET}/diffusion/PISA2022/student_questionnaire/CY08MSP_STU_QQQ.SAV"
with fs.open(FILE_PATH_S3, "rb") as s3f:
    with tempfile.NamedTemporaryFile(suffix=".sav") as tmp:
        tmp.write(s3f.read())
        tmp.flush()
        df = pd.read_spss(tmp.name)

print(df.head())
```

```python
df
```

```python
# On ne conserve que les observations associées à des élèves ayant à la fois
# répondu aux questionnaire sur la créativité (Option_CT), sur le bien-être (Option_WBQ)
# et dont les parents ont répondu au questionnaire (Option_PQ)

df_light=df[(df['Option_CT']=='Yes') & (df['Option_WBQ']=='Yes') & (df['Option_PQ']=='Yes')].copy()
```

```python
s=pd.Series(df.columns)
s
```

```python
## Dataset réduit à considérer à partir de maintenant
variables=list(s[s.str.startswith("ST") | s.str.startswith("PA") | s.str.startswith("CREAT")])
data=df_light[variables].copy()
data
```

# **Construction des Variables**

### **Variables explicatives**

#### **Variables du ST questionnaire** 

La variable *home*

Cette variable prend en compte le confort à la maison comme la présence ou non d'une chambre personnelle mais également l'accès ou non à des moyens de transport personnels (voiture, moto etc.). Cette variable ne prend pas en compte l'accès à la culture ni à des outils digitaux. 

Cette variable prend ses valeurs de 0 à 5

```python
home_liste=['ST250Q01JA']+['ST251Q0'+str(i)+'JA' for i in range(1,5)]
df_home=data[home_liste].copy()
df_home
```

La variable *digital_devices*

Cette variable prend en compte l'accessibilité à des outils digitaux comme des tablettes, des logiciels ou applications éducatives mais aussi l'accessibilité à un réseau internet. 

Cette variable prend ses valeurs de 0 à 4

```python
digital_liste=['ST250Q0'+str(i)+'JA' for i in range(2,6)]
df_digital=data[digital_liste].copy()
df_digital
```

La variable *culture*

Cette variable prend en compte l'accès à la culture en considérant notamment la présence ou non d'instruments de musique, d'oeuvres d'arts ou encore de livres. 

Cette variable prend ses valeurs de 0 à 3

```python
culture_liste=['ST251Q06JA','ST251Q07JA']
df_culture=data[culture_liste].copy()
df_culture
```

La variable *book*

Cette variable considère le nombre de livres présents à la maison. 

Cette variable prend ses valeurs de 0 à 6. 

```python
book_liste=['ST255Q01JA']
df_book=data[book_liste].copy()
df_book
```

La variable *siblings*

Cette variable prend en compte le nombre de frères et soeurs au sein du foyer de l'élève. 

Cette variable prend ses valeurs de 0 à 3.

```python
siblings_liste=['ST230Q01JA']
df_siblings=data[siblings_liste].copy()
df_siblings
```

La variable *parents*

Cette variable prend en compte le niveau d'étude et de qualification des deux parents.

Cette variable prend ses valeurs de 0 à 8.

```python
parents_liste=['ST005Q01JA','ST007Q01JA']+['ST006Q0'+str(i)+'JA' for i in range(1,6)]+['ST008Q0'+str(i)+'JA' for i in range(1,6)]
df_parents=data[parents_liste].copy()
df_parents
```

La variable *society*

Cette variable est associée à la classe sociale de la famille de l'élève. Cette variable à la fois en compte l'aspect financier mais également le niveau d'étude des parents ainsi que leurs qualifications et la position de leur métier au sein de la société. 

Cette variable prend ses valeurs de 0 à 10. 
0 : l'élève a du mal à s'intégrer, est rejeté
3 : l'élève sociabilise bien 

```python
society_liste=['ST259Q01JA']
df_society=data[society_liste].copy()
df_society
```

La variable *social*

Cette variable prend en compte l'intégration sociale de l'élève notamment à l'école. Pour cela, on considère sa facilité à se faire des amis ou non, s'il est harcelé etc. 

Cette variable prend ses valeurs de 0 à 3. 
0 : l'élève a du mal à s'intégrer, est rejeté
3 : l'élève sociabilise bien 

```python
social_liste=['ST034Q0'+str(i)+'TA' for i in range(1,7)]+['ST038Q0' + str(i) +'NA' for i in range(3,9)]+['ST038Q09JA','ST038Q10JA','ST038Q11JA']
# les variables ST034Q1 - 4 -6 JA ainsi que les variables de ST038 contribuent négativement à social
df_social=data[social_liste].copy()
df_social
```

La variable *safety*

Cette variable prend en compte la sécurité dans son environnement général, que ce soit à la maison, à l'école ou de façon générale à proximité de chez lui. 


Cette variable prend ses valeurs de 0 à 4.
0 : Tout va bien 
4 : Il est difficile d'aller à l'école en raison d'un cadré sécuritaire limité 

```python
safety_liste=['ST265Q0'+str(i)+'JA' for i in range(1,5)]+['ST266Q0'+str(i)+'JA' for i in range(1,6)]
# les variables de la deuxième liste contribuent négativement 
df_safety=data[safety_liste]
df_safety
```

La variable *school*

Cette variable est associée à l'environnement au sein de l'école notamment les conditions de travail des élèves et d'enseignement pour les professeurs. Cette variable prend en compte l'agitation des élèves, leur capacité de concentration mais aussi l'état des salles de cours et des écoles. 

Cette variable prend ses valeurs de 0 à 5.


```python
school_liste=['ST273Q0' + str(i) + 'JA' for i in range(1,8)]
df_school=data[school_liste]
df_school
```

La variable *teaching*

Cette variable est associé au niveau d'enseignement perçu par les élèves. Autrement dit, la qualité de l'enseignement reçu par l'élève selon sa perception (et donc sa satisfaction vis-à-vis des méthodes d'enseignement)

Cette variable prend ses valeurs de 0 à 4



```python
teaching_liste=['ST272Q01JA']+['ST270Q0'+str(i)+'JA' for i in range(1,5)]
df_teaching=data[teaching_liste].copy()
df_teaching
```

La variable *reasoning*

Cette variable est associé au niveau de tâches académiques qui nécessitent un certains raisonnement
mathématique par exemple. 


Cette variable prend ses valeurs de 0 à 4



```python
reasoning_liste=['ST285Q0' + str(i) + 'JA' for i in range(1,10)]+['ST275Q0' + str(i) + 'WA' for i in range(1,10)]+['ST276Q0' + str(i) + 'JA' for i in range(1,10)]+['ST276Q10JA']
df_reasoning=data[reasoning_liste].copy()
df_reasoning
```

La variable *family_support*

Cette variable mesure l'accompagnement de l'élève par sa famille. Autrement dit, elle permet de savoir si l'élève est soutenu ou non dans sa scolarité mais aussi de façon générale lors de périodes difficiles par sa famille.

Cette variable prend ses valeurs de 0 à 4

```python
family_support_liste=['ST300Q0' + str(i) + 'JA' for i in range(1,10)]+['ST300Q10JA']
df_family_support=data[family_support_liste]
df_family_support
```

La variable *school_support*

Cette variable mesure l'accompagnement de l'élève par l'école. Autrement dit, elle permet de savoir si l'élève est soutenu ou non dans sa scolarité mais aussi de façon générale lors de périodes difficiles par les équipes pédagogiques.  

Cette variable prend ses valeurs de 0 à 4

```python
school_support_liste=['ST267Q0' + str(i) + 'JA' for i in range(1,9)]+['ST354Q03JA','ST354Q08JA']+['ST348Q0' + str(i) + 'JA' for i in range(1,9)]
# Les variables ST267Q04JA et ST267Q08JA contribuent négativement à school_support
df_school_support=data[school_support_liste].copy()
df_school_support
```
